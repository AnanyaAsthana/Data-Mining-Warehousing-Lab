```python
import math
from collections import Counter

def entropy(data):
    counts = Counter(data)
    total = len(data)

    h = 0

    for count in counts.values():
        p = count / total
        h -= p * math.log2(p)

    return h

def joint_entropy(X, Y):
    pairs = list(zip(X, Y))
    counts = Counter(pairs)
    total = len(pairs)

    h = 0

    for count in counts.values():
        p = count / total
        h -= p * math.log2(p)

    return h

def conditional_entropy(X, Y):
    return joint_entropy(X, Y) - entropy(X)

def information_gain(X, Y):
    return entropy(Y) - conditional_entropy(X, Y)

def mutual_information(X, Y):
    return entropy(X) + entropy(Y) - joint_entropy(X, Y)

def gini_index(data):
    counts = Counter(data)
    total = len(data)

    gini = 1

    for count in counts.values():
        p = count / total
        gini -= p ** 2

    return gini

# Feature: Outlook
outlook = [
    "Sunny", "Sunny", "Overcast", "Rain", "Rain",
    "Rain", "Overcast", "Sunny", "Sunny", "Rain",
    "Sunny", "Overcast", "Overcast", "Rain"
]

# Target: Play Tennis
play = [
    "No", "No", "Yes", "Yes", "Yes",
    "No", "Yes", "No", "Yes", "Yes",
    "Yes", "Yes", "Yes", "No"
]

H_X = entropy(outlook)
H_Y = entropy(play)

H_XY = joint_entropy(outlook, play)

H_Y_given_X = conditional_entropy(outlook, play)

IG = information_gain(outlook, play)

MI = mutual_information(outlook, play)

Gini_X = gini_index(outlook)
Gini_Y = gini_index(play)

print("        INFORMATION THEORY CALCULATIONS")
print("\n1. Entropy")
print("----------------------------------------------")
print("Entropy of Outlook H(X) =", round(H_X, 4))
print("Entropy of Play H(Y)    =", round(H_Y, 4))

print("\n2. Joint Entropy")
print("----------------------------------------------")
print("H(X,Y) =", round(H_XY, 4))

print("\n3. Conditional Entropy")
print("----------------------------------------------")
print("H(Y|X) =", round(H_Y_given_X, 4))

print("\n4. Information Gain")
print("----------------------------------------------")
print("IG(X,Y) =", round(IG, 4))

print("\n5. Mutual Information")
print("----------------------------------------------")
print("I(X;Y) =", round(MI, 4))

print("\n6. GINI Index")
print("----------------------------------------------")
print("Gini Index of Outlook =", round(Gini_X, 4))
print("Gini Index of Play    =", round(Gini_Y, 4))
```
