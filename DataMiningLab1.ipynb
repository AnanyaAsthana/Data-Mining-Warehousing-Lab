```python
# Step 1: Import required libraries
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
import openpyxl # Added for reading .xlsx files

# Step 2: Upload dataset (CSV file)
from google.colab import files
uploaded = files.upload()

# Step 3: Load dataset
# Replace 'your_file.csv' with uploaded file name
df = pd.read_excel(list(uploaded.keys())[0]) # Changed to pd.read_excel

# Display original dataset
print("Original Dataset:\n")
print(df)

# Step 4: Separate numerical and categorical columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
cat_cols = df.select_dtypes(include=['object']).columns

# -------------------------------
# 🔹 Mean Imputation (Numerical)
# -------------------------------
mean_imputer = SimpleImputer(strategy='mean')
df_mean = df.copy()
df_mean[num_cols] = mean_imputer.fit_transform(df[num_cols])

print("\nDataset after Mean Imputation:\n")
print(df_mean)

# -------------------------------
# 🔹 Median Imputation (Numerical)
# -------------------------------
median_imputer = SimpleImputer(strategy='median')
df_median = df.copy()
df_median[num_cols] = median_imputer.fit_transform(df[num_cols])

print("\nDataset after Median Imputation:\n")
print(df_median)

# -------------------------------
# 🔹 Most Frequent (Categorical)
# -------------------------------
mode_imputer = SimpleImputer(strategy='most_frequent')
df_mode = df.copy()

# Apply to both numeric & categorical if needed
df_mode[num_cols] = mode_imputer.fit_transform(df[num_cols])
df_mode[cat_cols] = mode_imputer.fit_transform(df[cat_cols])

print("\nDataset after Most Frequent Imputation:\n")
print(df_mode)
```

```python
# Step 1: Import libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, Binarizer

# Step 2: Upload Excel file
from google.colab import files
uploaded = files.upload()

# Step 3: Load Excel file
file_name = list(uploaded.keys())[0]
df = pd.read_excel(file_name)

print("Original Dataset:\n")
print(df)

# Step 4: Select numerical columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns

# ----------------------------------------
# 🔹 Max-Min Normalization (0 to 1 scaling)
# ----------------------------------------
scaler = MinMaxScaler(feature_range=(0, 1))
df_minmax = df.copy()
df_minmax[num_cols] = scaler.fit_transform(df[num_cols])

print("\nDataset after Max-Min Normalization:\n")
print(df_minmax)

# ----------------------------------------
# 🔹 0/1 Normalization (Binarization)
# ----------------------------------------
# Threshold = 0.5 (you can change it)
binarizer = Binarizer(threshold=0.5)

df_binary = df.copy()

# Apply binarization only on numerical columns
df_binary[num_cols] = binarizer.fit_transform(df[num_cols])

print("\nDataset after 0/1 Normalization (Binarization):\n")
print(df_binary)

import matplotlib.pyplot as plt

# ----------------------------------------
# 📊 Graph 1: Original vs Max-Min Normalization
# ----------------------------------------

for col in num_cols:
    plt.figure(figsize=(10,4))

    plt.subplot(1,2,1)
    plt.plot(df[col], marker='o')
    plt.title(f"Original - {col}")
    plt.xlabel("Index")
    plt.ylabel(col)

    plt.subplot(1,2,2)
    plt.plot(df_minmax[col], marker='o', color='green')
    plt.title(f"Max-Min Normalized - {col}")
    plt.xlabel("Index")
    plt.ylabel(col)

    plt.tight_layout()
    plt.show()

# ----------------------------------------
# 📊 Graph 2: Histogram Comparison
# ----------------------------------------

for col in num_cols:
    plt.figure(figsize=(12,4))

    plt.subplot(1,2,1)
    plt.hist(df[col], bins=8)
    plt.title(f"Original Distribution - {col}")

    plt.subplot(1,2,2)
    plt.hist(df_minmax[col], bins=8)
    plt.title(f"Normalized Distribution - {col}")

    plt.tight_layout()
    plt.show()

# ----------------------------------------
# 📊 Graph 3: Original vs Binarized
# ----------------------------------------

for col in num_cols:
    plt.figure(figsize=(10,4))

    plt.subplot(1,2,1)
    plt.bar(range(len(df[col])), df[col])
    plt.title(f"Original - {col}")
    plt.xlabel("Index")
    plt.ylabel(col)

    plt.subplot(1,2,2)
    plt.bar(range(len(df_binary[col])), df_binary[col], color='red')
    plt.title(f"Binarized - {col}")
    plt.xlabel("Index")
    plt.ylabel("0/1")

    plt.tight_layout()
    plt.show()

# ----------------------------------------
# 📊 Graph 4: Boxplot Comparison
# ----------------------------------------

plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.boxplot([df[c] for c in num_cols], labels=num_cols)
plt.title("Original Data")

plt.subplot(1,2,2)
plt.boxplot([df_minmax[c] for c in num_cols], labels=num_cols)
plt.title("Max-Min Normalized Data")

plt.tight_layout()
plt.show()

# ----------------------------------------
# 📊 Graph 5: Heatmap Comparison
# ----------------------------------------

plt.figure(figsize=(8,6))
plt.imshow(df_minmax[num_cols].corr(), cmap='coolwarm', interpolation='nearest')
plt.colorbar()
plt.xticks(range(len(num_cols)), num_cols, rotation=45)
plt.yticks(range(len(num_cols)), num_cols)
plt.title("Correlation Heatmap (Normalized Data)")
plt.show()

# ----------------------------------------
# 🔹 Analysis (Comparison)
# ----------------------------------------
print("\n--- Analysis ---")
print("1. Max-Min normalization scales values between 0 and 1.")
print("2. It preserves the relative distribution of data.")
print("3. Binarization converts values into only 0 or 1.")
print("4. It depends on threshold (default used: 0.5).")
print("5. Useful for classification or presence/absence features.")
```
